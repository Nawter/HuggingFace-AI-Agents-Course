## Why Should You Use Agents?

Agents shine in scenarios where tasks are complex, dynamic, and involve external interactions. They're especially useful when workflows cannot be pre-programmed and require decision-making based on real-time data.

## When Not to Use Agents

If the task at hand can be handled using predefined workflows, agents may add unnecessary complexity. For instance, a customer support bot with clear pathways (like looking up trip details or contacting support) doesn't require an agent.

## When to Use Agents

For more nuanced interactions, agents are invaluable. Consider a user request like:

> "I'll arrive on Monday, but there might be a delay until Wednesday. Can you schedule a surfing lesson for Tuesday with an option to cancel?"

Such a query involves multiple variables—like checking weather forecasts, travel schedules, and booking availability—which a well-designed agent can handle efficiently by leveraging relevant tools.

## Key Features of SmolAgents

1. **Simplicity:** The library's core is just around 1,000 lines of code, making it easy to grasp and extend.
2. **Model Agnostic:** It supports any LLM, including OpenAI, Anthropic, and Hugging Face models, through the LiteLLM interface.
3. **Code Agents:** Unlike JSON-based tools, SmolAgents encourages using actual code for actions, improving adaptability and reuse.
4. **Hub Integration:** Tools can be shared and accessed via the Hugging Face Hub, fostering collaboration.

```python
import os
from dotenv import load_dotenv
from huggingface_hub import login, whoami

load_dotenv(dotenv_path=".env", override=True)
token = os.environ.get("HF_TOKEN")

# Validate before logging in
try:
    login(token=token, add_to_git_credential=False)
    user = whoami()
    print(f"✅ Logged in as: {user['name']}")
except Exception as e:
    print(f"❌ Login failed: {e}")
```

```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, InferenceClientModel

#Define an agent
agent = CodeAgent(tools=[DuckDuckGoSearchTool()], model=InferenceClientModel())

# Run a query
agent.run("How many seconds would it take for a snail at full speed to run through Pont des Arts?")
```

In this setup, the agent leverages the DuckDuckGo search tool to retrieve information and uses a Hugging Face API model for decision-making.

## Creating Custom Tools
Custom tools enhance your agent’s capabilities by allowing it to perform specialized tasks. Here’s how you can define one:

### Example: Travel Time Tool with Google Maps

```python
from dotenv import load_dotenv
load_dotenv()  # automatically reads the .env file
```

```python
from typing import Optional
from smolagents import CodeAgent, InferenceClientModel, tool
import googlemaps
import os


@tool
def get_travel_duration(start_location:str, destination_location:str, departure_time:Optional[int] = None) -> str:
    """
    Calculate the travel time between two locations using Google Maps.
    
    Args:
        start_location: The starting location for the trip.
        destination_location: The destination location for the trip.
        departure_time: The departure time as a Unix timestamp. Defaults to now if not provided.
    """    
   

    gmaps = googlemaps.Client(os.environ.get("GMAPS_API_KEY"))

    if not departure_time:
        from datetime import datetime
        departure_time =  datetime.now()

    directions = gmaps.directions(
        start_location,
        destination_location,
        mode="driving",
        departure_time=departure_time
    )

    return directions[0]["legs"][0]["duration"]["text"]

# Create an agent
agent = CodeAgent(tools=[get_travel_duration], model=InferenceClientModel())

# Run the agent
agent.run("Plan a trip from London to Manchester by car.")
```
